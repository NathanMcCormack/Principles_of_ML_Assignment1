```python
#looking for duplicate values - we would ignore any duplicate HR and respiration values as we are very likely to see duplicates in those fields, what we want to look for is if
#there are any duplicate ID && timestamp, to see if multiple values were entered for one timestamp

#using the pandas "subset" we can ignore other columns when looking for duplicates
df.duplicated(subset=["Participant", "Time(sec)"]).sum()
```

##Analysing Respiration and HR in a chosen participant

```python
#selecting participant 29 beacause they have the most data for respr < 7
p29 = df[df["Participant"] == 29]

#sort the readings in chronological order
p29 = p29.sort_values("Time(sec)")

#convert timestamps into seconds since their first recorded observation
time = p29["Time(sec)"] - p29["Time(sec)"].min()

#create the graph size
plt.figure(figsize=(12, 4))

plt.plot(time, p29["respr"])

plt.xlabel("Time since first observation (seconds)")
plt.ylabel("Respiration")
plt.title("Participant 29 - Respiration Over Time")

plt.show()
```

```python
plt.figure(figsize=(12, 4))

plt.scatter( #plots indiviudal data points
    time,
    p29["respr"],
    c=p29["Label"], #the labels will be coloured
    cmap="coolwarm",
    s=2 #size of plots
)

plt.xlabel("Time since first observation (seconds)")
plt.ylabel("Respiration")
plt.title("Participant 29 - Respiration and Stress Labels")

plt.colorbar(label="Stress Label (0 = Rest, 1 = Stress)") #legend for colour scale

plt.show()
```

```python
#Its interesting that the HR does not follow the same slopes as the respiration.

plt.figure(figsize=(12, 4))

plt.scatter( #plots indiviudal data points
    time,
    p29["HR"],
    c=p29["Label"], #the labels will be coloured
    cmap="coolwarm",
    s=2 #size of plots
)

plt.xlabel("Time since first observation (seconds)")
plt.ylabel("HR")
plt.title("Participant 29 - HR and Stress Labels")

plt.colorbar(label="Stress Label (0 = Rest, 1 = Stress)") #legend for colour scale

plt.show()
```
